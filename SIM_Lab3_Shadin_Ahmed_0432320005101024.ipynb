## Setup: Importing Necessary Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

# **Part 1**

```python
# Set a random seed for reproducibility
np.random.seed(42)

# 1. Generate 3,000 wait times using a uniform distribution between 2 and 12 minutes.
regular_day_wait_times = np.random.uniform(low=2, high=12, size=3000)

# 2. Generate 3,000 wait times using a normal distribution with Mean = 8, Std Dev = 2
busy_day_wait_times_raw = np.random.normal(loc=8, scale=2, size=3000)

# 3. Ensure no negative wait times (replace negatives with 0) for the normal distribution.
busy_day_times = np.maximum(0, busy_day_wait_times_raw)

print("Generated Uniform Wait Times (first 5):")
print(regular_day_wait_times[:5])
print("\nGenerated Normal Wait Times (first 5, after clipping):")
print(busy_day_times[:5])
```

# **Part 2**

```python
# Function to calculate and print key statistics for a given dataset
def calculate_statistics(data, name):
    mean_val = np.mean(data)
    median_val = np.median(data)
    std_dev_val = np.std(data)
    max_val = np.max(data)

    print(f"### {name} Statistics:")
    print(f"- Mean: {mean_val:.2f} minutes")
    print(f"- Median: {median_val:.2f} minutes")
    print(f"- Standard Deviation: {std_dev_val:.2f} minutes")
    print(f"- Maximum Wait Time: {max_val:.2f} minutes")

# Calculate statistics for Regular Day Wait Times
calculate_statistics(regular_day_wait_times, "Regular Day Wait Times (Uniform)")
print()

# Calculate statistics for Busy Day Wait Times
calculate_statistics(busy_day_times, "Busy Day Wait Times (Normal)")
```

# **Part 3**

```python
# Create a figure with two subplots for side-by-side comparison
plt.figure(figsize=(14, 6)) # Increased figure size for better readability

# Subplot 1: Regular Day Wait Times (Uniform Distribution)
plt.subplot(1, 2, 1) # 1 row, 2 columns, first plot
plt.hist(regular_day_wait_times, bins=20, edgecolor='black', alpha=0.8, color='skyblue')
plt.title('Regular Day Wait Times (Uniform Distribution)')
plt.xlabel('Wait Time (minutes)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.7)

# Subplot 2: Busy Day Wait Times (Normal Distribution)
plt.subplot(1, 2, 2) # 1 row, 2 columns, second plot
plt.hist(busy_day_times, bins=20, edgecolor='black', alpha=0.8, color='lightcoral')
plt.title('Busy Day Wait Times (Normal Distribution)')
plt.xlabel('Wait Time (minutes)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.7)

plt.tight_layout() # Adjust layout to prevent overlapping elements
plt.show()
```
